## 1. Colab シークレットに `OPENAI_API_KEY` を登録する

4. **値**に、事前配布された OpenAI API キーを貼り付けます。

## 2. 疎通確認セル（最重要・API は呼び出しません）

次のセルは、シークレット（または環境変数）から `OPENAI_API_KEY` を読み取り、
**キーが存在するかどうかだけ**を確認します。
**OpenAI API は呼び出しません。キーの値そのものも表示しません**（長さだけ表示します）。

- **成功**（✅）すれば、環境準備は完了です。
- **失敗**（❌）した場合は、上の手順 1 に戻り、シークレット名（`OPENAI_API_KEY`）と
  **[ノートブックからのアクセス]** が ON になっているかを確認してください。

```python
# 疎通確認: OPENAI_API_KEY の「存在」だけを確認する。
# ここでは OpenAI API は叩かない（存在チェックのみ）。キーの値そのものは表示しない。
import os

try:
    # Colab の場合: シークレットから読み込む
    from google.colab import userdata
    api_key = userdata.get("OPENAI_API_KEY")
except Exception:
    # Colab 以外（ローカルの Jupyter など）の場合: 環境変数から読み込む
    api_key = os.environ.get("OPENAI_API_KEY")

if api_key:
    # 存在確認 OK。値は出さず、長さ（文字数）だけ表示する
    print("✅ OPENAI_API_KEY を読み込めました (長さ:", len(api_key), "文字)")
    print("   ※ この演習では API は呼び出しません。実際の API 利用は第2章からです。")
else:
    print("❌ OPENAI_API_KEY が見つかりません。[シークレット] に登録し、ノートブックからのアクセスを有効にしてください。")
```

## 4. 公式ドキュメント検索ワークシート

ここからが、この演習の本題である**「一次情報を引く」体験**です。

### なぜ一次情報（公式ドキュメント）を引くのか

**この領域は変化がとても速い**からです。ブログ記事や書籍といった二次情報は、書かれた時点の
バージョンに固定されたまま古びていきます。判断に迷ったら、まず公式ドキュメントに当たる——
この習慣が、変化の速い技術を学び続ける力になります。

### 現行サイトと旧サイトの見分け方（URL のドメインで判断）

- **現行（最新）は `docs.langchain.com`** です。LangChain / LangGraph / Deep Agents / LangSmith の
  ドキュメントがこの 1 つのサイトに統合されています。
- **`python.langchain.com` はアーカイブ（旧版）**です。検索エンジン経由では今でも旧サイトに
  着地することが多いため、**必ず URL のドメインを確認**してください。これが新旧を見分ける最も確実な方法です。
- docs.langchain.com は Python と JavaScript の両方を扱います。**URL に `/oss/python/` が含まれているか**を
  確認する癖をつけましょう（Python 向けページである印です）。

### あなたのタスク

ブラウザで **https://docs.langchain.com** を開き、サイト内検索やナビゲーションを使って、

## 5. 3 つの URL の提出欄［解答記入済み］

下のセルに、3 つの URL を**記入済み**にしてあります。セルを実行すると、いずれも `docs.langchain.com`
ドメインなので **✅** になります。

> **重要な注記**: ここに記した URL は**参考**です。docs.langchain.com の URL 体系（パス構成）は
> 改訂されることがあります。**重要なのは、旧サイト `python.langchain.com` ではなく現行サイト
> `docs.langchain.com` で目的のページを見つけられること自体**です。受講者が辿り着いた URL が
> 下と多少違っていても、`docs.langchain.com` ドメインで該当ページに到達できていれば正解とします。
> 研修実施時は、講師がその場で docs.langchain.com を開いて最新の URL を最終確認してください。

```python
# 検索した URL（解答・記入済み。すべて docs.langchain.com ドメイン）
create_agent_url = "https://docs.langchain.com/oss/python/langchain/agents"            # create_agent の解説ページ（Agents）
middleware_list_url = "https://docs.langchain.com/oss/python/langchain/middleware/built-in"  # Middleware 組み込み一覧（Prebuilt middleware）
changelog_url = "https://docs.langchain.com/oss/python/releases/changelog"             # リリース changelog

# 簡易チェック: 3 つとも docs.langchain.com ドメインか確認する
for name, url in [("create_agent", create_agent_url), ("middleware", middleware_list_url), ("changelog", changelog_url)]:
    ok = url.startswith("https://docs.langchain.com")
    print(("✅" if ok else "⚠️ "), name, ":", url or "(未記入)")
```
